```python
import pandas as pd


data = {
    'Name': ['Alice', None, 'Charlie', 'Alice'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', None, 'Paris', 'LA']
}

data2 = {
    'Name': ['Bob', 'Charlie', None],
    'Age': [25, 30, 35],
    'Place': ['New York', 'London', 'Paris']
}

df = pd.DataFrame(data)
c = pd.Series()
df2 = pd.DataFrame(data2)

print(df["Name"].dtype)
print(c.add(df2["Name"].value_counts(), fill_value = 0))

print(df2.isna().sum() + df.isna().sum())
print(df2.isna().sum().add(df.isna().sum(), fill_value = 0))

```
